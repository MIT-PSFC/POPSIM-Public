# Power balance
This notebook demonstrates using the power balance module.

```python
%load_ext autoreload
%autoreload 2
```

```python
from popsim.modules.power_balance import PowerBalance
from popsim.simulate import make_time_base, simulate, SimInput
from popsim.visualize import visualize_time_series

# Set initial state
state = PowerBalance.State(
    stored_energy=0.0,
)

# Set params
tau_E = 1.0  # Energy confinement time [s]

params = PowerBalance.Params(
    P_aux={0.1: 0.0, 0.99: 0.0, 1.0: 1.0, 9.0: 1.0},  # Auxiliary power [MW]
    confinement_time=tau_E,  # energy confinement time in seconds
)
```

```python
time_base = make_time_base(t0=0.0, t1=10.0, dt=0.01)
power_balance_module = PowerBalance(config=PowerBalance.Config())
sim_input = SimInput(time=time_base, initial_state=state, params=params)
sol = simulate(power_balance_module, sim_input)
```

```python
sol_xarray = sol

visualize_vars = [
    "state.stored_energy",
]
visualize_time_series(sol_xarray[visualize_vars])
```
