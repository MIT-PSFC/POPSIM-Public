# ICRH zone
This notebook demonstrates using the icrh zone module.

```python
%load_ext autoreload
%autoreload 2
```

```python
from popsim.modules.icrh_zone import IcrhZone
from popsim.simulate import SimInput, make_time_base, simulate
from popsim.visualize import visualize_time_series

# Set initial state
state = IcrhZone.State()

# Set inputs
inputs = IcrhZone.Inputs(
    frequency_command=120,
    power_command={0.0: 0.0, 1.0: 2e6},
)
```

```python
time_base = make_time_base(t0=0.0, t1=10.0, dt=0.01)
icrh_zone_module = IcrhZone(config=IcrhZone.Config())
sim_input = SimInput(time=time_base, initial_state=state, inputs=inputs)
sol = simulate(icrh_zone_module, sim_input)
```

```python
sol_xarray = sol

visualize_vars = [
    "output.reflected_power",
    "output.transmitted_power",
]
visualize_time_series(sol_xarray[visualize_vars])
```
