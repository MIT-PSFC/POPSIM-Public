# Comparing Against Creely, et al. 2020

This notebook provides a comparison of `TdPopcon` simulation results from the values reported in Creely, et al. 2020. 

There is a unit test in `test_td_popcon` that will automatically run a simulation and compare the results and make sure they are within some specified tolerance (go take a look!). Here, we are using the exact same code, but then post-processing the results to visualize.

```python
from popsim.tests.test_modules.test_td_popcon import generate_sim_and_checks
from popsim.visualize import visualize_time_series
from popsim.xarray_utils import time_and_pytree_to_xarray

check_dictionary, sol, ts = generate_sim_and_checks(only_return_final=False)

sim_results = {k: v["sim_result"] for k, v in check_dictionary.items()}
expected = {k: v["creely2020"] for k, v in check_dictionary.items()}


ds = time_and_pytree_to_xarray(ts, sim_results, multi_simulation=False)

visualize_time_series(ds, hlines=expected)
```
