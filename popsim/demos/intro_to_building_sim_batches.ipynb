# Introduction to popsim.simulate

`popsim.simulate` provides a common interface for **all** simulation modules and provides a couple of features:

1. **Time-Dependent Parameters**: the ability to specify any element of the `Params` structure as a function of time.
2. **Running Batches of Simulations**: by simply providing a list of `SimInput` objects.
3. **Utilities to Build Batches of Simulations**
