```python
from popsim.modules.icrh_zone import IcrhZone
import typing

import equinox as eqx
from jaxtyping import Array
import jax.numpy as jnp
import jax
import diffrax
from popsim.interp import resolve_paths
from popsim.xarray_utils import solution_to_xarray
from popsim.visualize import visualize_time_series
from popsim.param_utils import build_vectorized_configs
import popsim.param_utils as param_utils
from popsim.simulate import simulate

# Set initial state
state = IcrhZone.State()

# Set params
params = IcrhZone.Params(
    frequency_command = 120,
    power_command = {0.0:0.0, 1.0:2e6},
)
```

```python
time_base = param_utils.make_time_base(t0=0.0, t1=10.0, dt=0.01)
icrh_zone_module = IcrhZone(config=IcrhZone.Config())
sol = simulate(icrh_zone_module, time_base, state, params)
```

```python
sol_xarray = sol

visualize_vars = [
    "aux.reflected_power",
    "aux.transmitted_power",
]
visualize_time_series(sol_xarray[visualize_vars])
```
