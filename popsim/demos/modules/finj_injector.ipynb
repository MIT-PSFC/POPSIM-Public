# FINJ injector
This notebook demonstrates using the FINJ injector module.

```python
%load_ext autoreload
%autoreload 2
```

```python
from popsim.modules.finj_injector import FinjInjector
from popsim.simulate import make_time_base, simulate
from popsim.visualize import visualize_time_series

# Set up time constants
valve_flow_rate_tau = 2.0  # valve flow rate confinement time [s]
pipe_flow_rate_tau = 3.0  # pipe flow rate confinement time [s]

# Set initial state
state = FinjInjector.State(
    valve_flow_rate=2.0,
    pipe_flow_rate=1.0,
)

# Set params
params = FinjInjector.Params(
    flow_rate_command=3.0,  # [Pa m^3/s]
    valve_flow_rate_tau=valve_flow_rate_tau,  # [s]
    pipe_flow_rate_tau=pipe_flow_rate_tau,  # [s]
)
```

```python
time_base = make_time_base(t0=0.0, t1=10.0, dt=0.01)
finj_injector_module = FinjInjector(config=FinjInjector.Config())
sol = simulate(finj_injector_module, time_base, state, params)
```

```python
sol_xarray = sol

visualize_vars = [
    "state.valve_flow_rate",
    "state.pipe_flow_rate",
]
visualize_time_series(sol_xarray[visualize_vars])
```
