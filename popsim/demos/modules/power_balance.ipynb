```python
from popsim.modules.power_balance import PowerBalance
import typing

import equinox as eqx
import jax.numpy as jnp
import jax
from jaxtyping import Array
import diffrax
from popsim.interp import resolve_paths
from popsim.xarray_utils import solution_to_xarray
from popsim.visualize import visualize_time_series
from popsim.param_utils import build_vectorized_configs
import popsim.param_utils as param_utils
from popsim.simulate import simulate

# Set initial state
state = PowerBalance.State(
    stored_energy=0.0,
)

# Set params
tau_E = 1.0 #Energy confinement time [s]

params = PowerBalance.Params(
    P_aux={0.1:0.0, 0.99:0.0, 1.0:1.0, 9.0:1.0},  # Auxiliary power [MW]
    confinement_time=tau_E,  # energy confinement time in seconds
)
```

```python
time_base = param_utils.make_time_base(t0=0.0, t1=10.0, dt=0.01)
power_balance_module = PowerBalance(config=PowerBalance.Config())
sol = simulate(power_balance_module, time_base, state, params)
```

```python
sol_xarray = sol

visualize_vars = [
    "state.stored_energy",
]
visualize_time_series(sol_xarray[visualize_vars])
```
